## Zadanie: wybrać takie wiersze, aby q6 > percentyl: 75% lub mniejsza od pertentyl: 25%

```python
(1 == 0) or (1 == 1)  #  | operator dla warunku LUB w pandas
```

```python
q_25 = df.describe().loc['25%', 'q6']  # ustalam ile wynosi q25
q_75 = df.describe().loc['75%', 'q6']  # ustalam ile wynosi q75
```

```python
q_25, q_75
```

```python
mask_1 = df['q6'] <= q_25 #  tworzę maskę
mask_2 = df['q6'] >= q_75 #  tworzę maskę
df_q25_75 = df[mask_1 | mask_2]  # tworzę data frame dla wybranego warunku LUB
df_q25_75
```

## Korzystając z operatora `&` będą realizował warunek `and`

```python
for ind, row in df_q25_75.iterrows():
  # print(ind)
  print(row['q1'])  # to będzie wartość dla zmiennej q1 i pierwszego wiersza
  break
```
