# Zadanie: utworzyć wykres scatter dla krajów kiedy wsp. korelacji jest większy niż 0.9.

```python
df_trans = df_transpose.copy()
```

```python
df_corr.index, df_corr.columns
```

```python
range(len(df_corr))
```

```python
df_trans = df_transpose_standaryzacja.copy()
```

```python
df_corr = df_transpose_standaryzacja.corr()
df_corr
```

```python
for row_id in range(len(df_corr)):
  for col_id in range(len(df_corr)):
    row = df_corr.index[row_id]
    col = df_corr.columns[col_id]
    if df_corr.loc[row, col] >= 0.99 and df_corr.loc[row, col] != 1 and row_id > col_id:
      print(f"Wiersz: {row}, Kolumna: {col}, Wartość: {df_corr.loc[row, col]}")
      x = df_trans[row]
      y = df_trans[col]
      corr = df_trans[[row,col]].corr()
      corr = corr.iloc[0,1]
      plt.scatter(x,y)
      plt.title(f"Wykres rozproszenia: {row} i {col} {corr.round(2)}")
      plt.xlabel(row)
      plt.ylabel(row)
      plt.show()
```
